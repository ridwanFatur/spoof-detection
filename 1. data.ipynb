```python
def draw_bbox(item, color="red", width=2):
    image = item["Filepath"].copy()
    bbox = item["Bbox"]

    draw = ImageDraw.Draw(image)
    draw.rectangle(bbox, outline=color, width=width)

    return image
```

```python
draw_bbox(item, color="red", width=2)
```
